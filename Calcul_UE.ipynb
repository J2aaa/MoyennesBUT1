```python
import tkinter as tk
from tkinter import ttk, GROOVE
import numpy as np

fen = tk.Tk()
fen.geometry("800x700")
fen.title("Calcul des moyennes")

frm = ttk.Frame(fen)
frm.grid(padx=20, pady=5)


matieres_noms = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

titre = ttk.Label(frm, text="Saisie des notes", font=("Arial", 14, "bold"))
titre.grid(column=0, row=0, columnspan=2, pady=5)

saisies = []


for i, nom in enumerate(matieres_noms):
    mat = ttk.Label(frm, text=nom)
    mat.grid(column=0, row=i+1, padx=10, pady=5, sticky="w")
    
    data = ttk.Entry(frm)
    data.grid(column=1, row=i+1, padx=10, pady=5)
    saisies.append(data) 


coef1=[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2=[4,0,2,8,6,0,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3=[4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

def recup():
    notes = [float(data.get()) for data in saisies]
    UE1 = ttk.Label(frm, text="Moyenne de l'UE1 :")
    UE1.grid(column=2, row=3, padx=20)
    moy1 = ttk.Label(frm, text=str(round(np.average(notes, weights=coef1), 3)), relief=GROOVE, padding=5)
    moy1.grid(column=4, row=4)
    
    UE2 = ttk.Label(frm, text="Moyenne de l'UE2 :")
    UE2.grid(column=2, row=5, padx=20)
    moy2 = ttk.Label(frm, text=str(round(np.average(notes, weights=coef2), 3)), relief=GROOVE, padding=5)
    moy2.grid(column=4, row=10)

     
    UE3 = ttk.Label(frm, text="Moyenne de l'UE3 :")
    UE3.grid(column=2, row=7, padx=20)
    moy3 = ttk.Label(frm, text=str(round(np.average(notes, weights=coef3), 3)), relief=GROOVE, padding=5)
    moy3.grid(column=4, row=15)
        

btn_calculer = ttk.Button(frm, text="Calculer les moyennes", command=recup)
btn_calculer.grid(column=4, row=1)


fen.mainloop()
```

```python
from tkinter import *
from tkinter import ttk
import numpy as np
import matplotlib.pyplot as plt

fen = Tk()
fen.title("Résultats du semestre 1")
frm = ttk.Frame(fen)
frm.grid()

matieres = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]
saisies = []
coef1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

for i, matiere in enumerate(matieres):
  mat = ttk.Label(frm, text=matiere)
  mat.grid(column=0, row=i+1)
  data = ttk.Entry(frm)
  data.grid(column=1, row=i+1)
  saisies.append(data)

def recup():
  notes = [float(data.get()) for data in saisies]

  UE1 = ttk.Label(frm, text="Moyenne de l'UE1")
  UE1.grid(column=2, row=3)
  moy1 = ttk.Label(frm, text=round(np.average(notes, weights=coef1), 3))
  moy1.grid(column=3, row=3)

  UE2 = ttk.Label(frm, text="Moyenne de l'UE2")
  UE2.grid(column=2, row=5)
  moy2 = ttk.Label(frm, text=round(np.average(notes, weights=coef2), 3))
  moy2.grid(column=3, row=5)

  UE3 = ttk.Label(frm, text="Moyenne de l'UE3")
  UE3.grid(column=2, row=7)
  moy3 = ttk.Label(frm, text=round(np.average(notes, weights=coef3), 3))
  moy3.grid(column=3, row=7)

btn_val = ttk.Button(frm, text="Valider la saisie", command=recup)
btn_val.grid(column=2, row=1)

def code_couleur(moy) :
    if moy >= 10 :
        return "green"
    elif moy <8 :
        return "red"
    else :
        return "orange"
moyS1 = [$moy1,$moy2,$moy3]
colors = [code_couleur(moy) for moy in moyS1]

plt.bar(["UE1","UE2","UE3"],moyS1, color = colors)
plt.show

fen.mainloop()
```
